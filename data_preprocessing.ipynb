```python
df["ACTIVE_MEMBER"].isnull().sum()
```

```python
from scipy import stats

data = df['ACTIVE_MEMBER'].dropna()
mean_val, median_val, skew_val = data.mean(), data.median(), data.skew()

fig, ax = plt.subplots(figsize=(9, 5))

# Actual distribution (KDE)
sns.kdeplot(data, ax=ax, color='#2c7fb8', linewidth=2, label='ACTIVE_MEMBER (actual)', fill=True, alpha=0.3)

# Normal distribution with same mean/std, for comparison -> shows how far it deviates from symmetric
x = np.linspace(data.min(), data.max(), 200)
normal_curve = stats.norm.pdf(x, mean_val, data.std())
ax.plot(x, normal_curve, color='gray', linestyle='--', linewidth=2, label='Normal curve (same mean/std)')

ax.axvline(mean_val, color='red', linestyle='--', label=f'Mean = {mean_val:.2f}')
ax.axvline(median_val, color='black', linestyle='-', label=f'Median = {median_val:.2f}')

ax.set_xlim(0, 15)
ax.set_title(f'ACTIVE_MEMBER Skewness = {skew_val:.2f} (Right-Skewed)')
ax.set_xlabel('Active Member Count')
ax.set_ylabel('Density')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
skew_val = df['ACTIVE_MEMBER'].skew()

if skew_val > 0.5:
    direction = "Right-skewed (positive skew) -> tail stretches toward higher values -> use MEDIAN for imputation"
elif skew_val < -0.5:
    direction = "Left-skewed (negative skew) -> tail stretches toward lower values -> use MEDIAN for imputation"
else:
    direction = "Approximately symmetric -> MEAN is also a reasonable choice"

print(f"ACTIVE_MEMBER skewness: {skew_val:.3f}")
print(direction)
```
